```python
import os
from pathlib import Path
from operator import itemgetter
from collections import Counter

import math
import random

import json
import csv
import strictyaml as syaml

import numpy as np
import cv2

import matplotlib.pyplot as plt
from IPython import display as ipd
from tqdm import tqdm

import bootleg_utils
import bbox_utils

np.set_printoptions(threshold=30_000)

def imshow(img, cmap='gray', axis=True):
  axis = 'on' if axis else 'off'
  plt.imshow(img, cmap=cmap)
  plt.axis(axis)
  plt.show()
```

```python
root_dir = Path('.')
data_dir = root_dir / 'data'
score_dir = root_dir / 'scores'

with open(data_dir / 'scores.yaml') as f:
  score_metadata = syaml.load(f.read())

score_metadata = score_metadata.data

tgt_list = [
  ( path, path.parent.parent / 'flattened' / path.name )
  for mscore_id, infos in score_metadata.items()
  for path in list(sorted((score_dir / infos['path'] / 'images' / 'original').glob('*.png')))
][:35]
```

```python
for src, tgt in tgt_list:
  print(src)
  
  og = cv2.imread(src, cv2.IMREAD_UNCHANGED)
  plt.figure(figsize=(20, 10))
  imshow(og)
  
  img = cv2.imread(tgt, cv2.IMREAD_UNCHANGED)
  plt.figure(figsize=(20, 10))
  imshow(img[..., 0])
  print('\n')
```
